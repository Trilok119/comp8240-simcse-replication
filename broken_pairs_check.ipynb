Re-computes the Spearman correlation (×100) between each method's score and the automatic level on the 400 WikiSim pairs, with and without the pairs that contain a broken sentence (same `is_broken` rule as `SimCSE_replication.ipynb`, Section 3.6). Needs only pandas and scipy; run it from the repository root. These are the numbers quoted in the report's *Data quality* paragraph.

```python
import re, pandas as pd
from scipy.stats import spearmanr

d400 = pd.read_csv("data/wikisim_pairs.csv")

def is_broken(s):
    return s.count("(") != s.count(")") or re.search(r"\b(i\.e|e\.g|c|ca|approx|St|Dr|Mt|vs|No)\.$", s) is not None

bad = d400.s1.apply(is_broken) | d400.s2.apply(is_broken)
keep = d400[~bad]
print("Broken pairs removed:", bad.sum(), "| pairs left:", len(keep))
print(f"{'method':14s} {'all 400':>8s} {'without':>8s}")
for c in ["word_overlap", "unsup_simcse", "sup_simcse"]:
    print(f"{c:14s} {spearmanr(d400.level, d400[c]).correlation*100:8.2f} {spearmanr(keep.level, keep[c]).correlation*100:8.2f}")
```
